```python
import numpy as np
from dezero import Variable
import dezero.functions as F

x = Variable(np.array([[1,2,3],[4,5,6]]))
y = F.get_item(x, 1)
y
```

```python
y.backward()
print(x.grad)
```

```python
from dezero.models import MLP

model = MLP((10, 3))
x = np.array([[0.2, -0.4]])
y = model(x)
print(y)
```
