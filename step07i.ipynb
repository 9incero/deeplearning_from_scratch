```python
from steps import Square, Exp, Function, Variable


A = Square()
B = Exp()
C = Square()
```

```python

import numpy as np

x = Variable(np.array(0.5))

a = A(x)
b = B(a)
y = C(b)


assert y.creator == C
assert y.creator.input == b
assert y.creator.input.creator == B
assert y.creator.input.creator.input == a

y.grad = np.array(1.0)

C = y.creator
b = C.input
b.grad = C.backward(y.grad)

B = b.creator
a = B.input
a.grad = B.backward(b.grad)

A = a.creator
x = A.input
x.grad = A.backward(a.grad)

x.grad
```
